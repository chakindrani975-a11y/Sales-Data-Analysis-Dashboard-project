```python
df["saledate"] = pd.to_datetime(df["saledate"], errors="coerce")
```

```python
df["saledate"].dtype
```

```python
df["saledate"].isnull().sum()
```

```python
df["sale_year"] = df["saledate"].dt.year
```

```python
df[["saledate", "sale_year"]].head()
```

```python
df["vehicle_age"] = (
    df["sale_year"] - df["year"]
)
```

```python
df[["year", "sale_year", "vehicle_age"]].head()
```

```python
(df["vehicle_age"] < 0).sum()
```

```python
df["price_difference"] = (
    df["sellingprice"] - df["mmr"]
)
```

```python
df[
    ["mmr", "sellingprice", "price_difference"]
].head()
```

```python
df["price_difference_pct"] = (
    (df["sellingprice"] - df["mmr"])
    / df["mmr"]
) * 100
```

```python
df[
    ["mmr", "sellingprice", "price_difference_pct"]
].head()
```

```python
df["sale_month"] = df["saledate"].dt.month
```

```python
df["sale_month_name"] = (
    df["saledate"].dt.strftime("%B")
)
```

```python
df["sale_quarter"] = (
    "Q" + df["saledate"].dt.quarter.astype(str)
)
```

```python
df[
    [
        "year",
        "saledate",
        "sale_year",
        "vehicle_age",
        "mmr",
        "sellingprice",
        "price_difference",
        "price_difference_pct",
        "sale_month",
        "sale_month_name",
        "sale_quarter"
    ]
].head(10)
```

```python
categorical_columns = [
    "make",
    "model",
    "trim",
    "body",
    "transmission",
    "color",
    "interior"
]

for column in categorical_columns:
    df[column] = df[column].fillna("Unknown")
```

```python
df[categorical_columns].isnull().sum()
```

```python
df = df.dropna(
    subset=["sellingprice"]
)
```

```python
df["sellingprice"].isnull().sum()
```

```python
df = df.dropna(
    subset=["saledate"]
)
```

```python
df["saledate"].isnull().sum()
```

```python
df_public = df.drop(
    columns=["vin"]
)
```
